# Clase 15 — Trabajo en el proyecto

**Módulo:** Ciclo MLOps

**Intención:** avanzar el proyecto hasta tener modelos evaluados y registrados,
un pipeline de entrenamiento y una API conectada con una interfaz.

## Actualizar el material del curso

Desde la raíz del repositorio público, revisa `git status`. Si estás en
`main` y no aparecen cambios locales, ejecuta `git pull`. Si hay cambios,
conserva tu trabajo y pide orientación antes de continuar.
El trabajo del proyecto se realiza en el repositorio privado del equipo.

**Resultado observable:** avance funcional en el propio proyecto, con

| Componente | Resultado esperado |

## 3. Trabajo en el proyecto

Apliquen las correcciones anteriores en los notebooks, informe, código y
documentación afectados. Comprueben los resultados; no se solicita una
bitácora ni un documento de seguimiento.

Continúen los componentes pendientes de la tarea. Distribuyan responsabilidades
entre modelado/pipeline, API e interfaz y revisen las conexiones entre sus partes.
Mantengan `notebooks/03-experimentos.ipynb`, `src/entrenamiento/`,
`src/api/` y `src/frontend/`.
Cada componente tiene su notebook técnico detallado, según la tarea; el
informe reúne los hallazgos principales y referencia esos cuadernos.

Una sola persona ejecuta los experimentos remotos y registra los modelos desde
su cuenta personal de Databricks. Todo el equipo participa en el desarrollo y
análisis. No compartan contraseñas, tokens ni archivos `.env`.

Para discutir una duda, muestren el archivo o resultado, lo que esperaban,
lo observado y lo que intentaron.

**Checkpoint:** el equipo puede mostrar una corrección o un componente que
avanzó, ejecutar su comprobación y explicar cómo se relaciona con el proyecto.

## 4. Experimentos y modelos del equipo

Utilicen un identificador descriptivo sin datos personales. La convención es:

| Recurso | Nombre |
|---|---|
| Experimento remoto | `/Shared/pcd-otono-2026-proyecto-<proyecto-en-kebab-case>` |
| Modelo registrado | `workspace.default.pcd_otono_2026_<proyecto_en_snake_case>` |

El experimento es una ruta de Workspace; el modelo es un nombre de tres
niveles de Unity Catalog. El catálogo es `workspace` y el esquema `default`.
Sustituyan sólo el identificador del proyecto y utilicen puntos entre
catálogo, esquema y modelo.

Por ejemplo, `abandono-clientes` se convierte en
`/Shared/pcd-otono-2026-proyecto-abandono-clientes` y
`workspace.default.pcd_otono_2026_abandono_clientes`.

El experimento reúne las familias; cada una tiene su parent run de HPO y
sus trials. Los mejores pipelines se registran como versiones de un mismo
modelo y `champion` apunta al seleccionado.
Las reglas completas y ejemplos de runs están en la tarea.

**Checkpoint:** todos reconocen la ruta del experimento, el nombre del modelo
y qué persona ejecutará en el workspace personal.

## 5. Comprobar y conservar evidencia

Ejecuten los componentes que vayan completando y guarden capturas legibles en
`evidencia/`, enlazadas desde el informe o README. La tarea precisa los casos:
experimento remoto, HPO por familia, comparación, Registry, flow de Prefect,
API válida e inválida y frontend consumiendo la API.

Las imágenes deben mostrar acciones y resultados, no sólo código. Conserven
visibles los nombres del experimento/modelo y las métricas; oculten credenciales,
correo y URL personal del workspace.

Revisen cambios y evidencias entre integrantes. Usen la rama
`feat/segunda-entrega` y el [flujo de PR del curso](../docs/flujo-tareas.md).
Si hay trabajo pendiente en otra rama, coordinen su integración antes de
cambiar de rama o iniciar la entrega.

## Errores frecuentes

| Problema | Qué revisar |
|---|---|
| Tres modelos con HPO que son la misma familia. | Elegir algoritmos de tres familias y definir una búsqueda por cada una. |
| La evaluación final participa en el tuning. | Separar datos y comprobar las entradas de la función objetivo. |
| El artefacto no transforma las entradas como al entrenar. | Conservar y cargar el mismo preprocesamiento con el modelo. |
| La API devuelve otra versión. | Comprobar la referencia cargada y la versión asociada a `champion`. |
| La interfaz muestra un resultado sin llamar a la API. | Revisar la solicitud HTTP y relacionarla con la respuesta y el formulario. |
| Las capturas no permiten revisar las corridas. | Mostrar ruta, trials, parámetros, métricas y versiones con texto legible. |

## Recopilación

El proyecto conecta decisiones de datos y evaluación con un modelo
versionado, un entrenamiento reproducible y una predicción accesible.
Las correcciones se incorporan al trabajo existente; las capturas y la
interpretación permiten comprobar lo realizado.

La segunda entrega completa ese recorrido local. El informe explica
decisiones y resultados y el README indica cómo repetir la ejecución.
