## 4. 📦 Preparar la versión desplegable

Mantengan Python 3.12, dependencias y lock coherentes, `.env.example` sin secretos y Dockerfiles llamados `Dockerfile`. Documenten los contextos de construcción y la red host usada en el recorrido del curso.

Publiquen la rama para probarla en EC2, incorporen las evidencias y completen el merge según la tarea. No suban datos, modelos, llaves, tokens ni `.env`.


## 5. ☁️ Una instancia para el proyecto del equipo

Desplieguen en **la EC2 ya asignada a uno de los integrantes**, en `us-east-1`. No creen una instancia por integrante. Acuerden una persona responsable de administrar EC2 y conserven separados los archivos/servicios del laboratorio, examen y proyecto. Cada responsable utiliza sus propios accesos; no compartan llaves o contraseñas.

Usen el procedimiento conocido de la [Clase 20](clase-20-aws-ec2-compose.ipynb): conectar por SSH, obtener el repositorio privado, crear configuración externa, construir/arrancar Compose y comprobar la predicción. Si el modelo proviene de la cuenta Databricks de otra persona del equipo, coordinen un artefacto propio confiable con procedencia/versión, sin compartir tokens ni versionarlo.

> **SSH y nueva IP:** si la IP pública de la computadora administradora cambió, actualicen el origen de la regla SSH (TCP 22) del grupo de seguridad a **My IP**. No abran SSH a todo Internet. Comprueben la dirección pública actual de EC2 antes de conectar.

Comprueben el acceso al frontend desde otra computadora, la respuesta y la versión. Revisen logs y detengan los servicios y la instancia al terminar de verificar. No terminen la instancia reutilizada sólo por entregar. Si falta acceso o recursos, avisen al profesor con el diagnóstico y continúen con las comprobaciones locales.

**Checkpoint:** otra computadora obtiene una predicción y el equipo puede relacionarla con el modelo evaluado y la versión desplegada.

## Errores frecuentes

| Síntoma | Qué revisar |
|---|---|
| Local funciona, contenedor no. | Contexto, rutas, dependencias, archivos ignorados y configuración externa. |
| UI no conecta con API. | API lista, dirección/puerto, red host y entorno dentro del contenedor. |
| EC2 no permite SSH. | Instancia activa, destino actual, llave correcta y origen autorizado para la IP de la computadora. |
| Aplicación remota usa otro modelo. | Procedencia del artefacto, versión cargada y relación con Registry/evaluación. |
| Otro servicio ocupa recursos. | Identificar servicios propios; coordinar detención y nombres sin borrar trabajo o detener recursos ajenos. |
| Capturas no coinciden con el código entregado. | Actualizar EC2 tras integrar cambios y repetir los casos afectados. |


## 🧩 Recopilación

Una aplicación end to end conecta datos y evaluación con un modelo versionado, una API, una interfaz y un entorno ejecutable. El despliegue conserva esas decisiones: cambiar de computadora no debe cambiar el procesamiento ni ocultar qué versión responde.

Las correcciones se aplican al proyecto existente y se validan antes de ampliar. El informe interpreta, los notebooks técnicos respaldan y el README permite reproducir.
