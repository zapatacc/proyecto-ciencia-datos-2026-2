## 2. Terminal, shell y CLI

Estos términos están relacionados, pero no significan exactamente lo mismo:

- la **terminal** es la aplicación o panel donde escribes y ves resultados;
- el **shell** es el programa que recibe el texto, interpreta su sintaxis, localiza programas y muestra su salida;
- la **CLI** (*Command-Line Interface*) es la forma de interactuar mediante texto;
- **Git Bash** instala Bash y herramientas tipo Unix junto con Git en Windows.

Existen distintos programas que funcionan como shell:

- **Bash:** es común en Linux, servidores y Git Bash.
- **Zsh:** es el shell predeterminado en versiones modernas de macOS y tiene una sintaxis muy cercana a Bash.
- **Fish:** prioriza una experiencia interactiva amigable y utiliza algunas reglas diferentes de Bash.
- **CMD:** es el shell clásico de Windows.
- **PowerShell:** es un shell de Microsoft que utiliza comandos y tuberías orientados a objetos de .NET.

Todos son válidos en su contexto, pero no siempre aceptan exactamente la misma sintaxis.

En Windows usaremos **Git Bash** porque proporciona los mismos comandos básicos tipo Unix que encontraremos después en Linux, servidores y contenedores. Así podemos compartir instrucciones como `pwd`, `ls`, `cp` y `grep` entre Windows, macOS y Linux. Git Bash no convierte Windows en Linux: es un entorno compatible que representa, por ejemplo, `C:\Users\Ana` como `/c/Users/Ana`. En macOS puede usarse Zsh; para los comandos de esta clase se comportará de forma equivalente.

Al abrir el shell aparece un *prompt* parecido a éste:

```text
ana@EQUIPO MINGW64 ~/cursos
$
```

El prompt indica que el shell está listo. Puede mostrar usuario, equipo, shell y ruta actual. No copies `ana@EQUIPO`, `MINGW64` ni `$`: escribe únicamente el comando después del prompt.

Un comando puede incluir el nombre de un programa, opciones y argumentos. Observa un ejemplo que sí puedes ejecutar desde la raíz del repositorio del curso:

```bash
ls -l -h labs
```

| Parte | En el ejemplo | Papel |
|---|---|---|
| Programa | `ls` | Lista archivos y carpetas. |
| Opciones o *flags* | `-l -h` | Cambian la presentación: `-l` solicita detalles y `-h` muestra tamaños legibles. |
| Argumento | `labs` | Indica **sobre qué carpeta** debe trabajar `ls`. Es una ruta del repositorio, no una palabra especial del comando. |

Un **argumento** es un dato que entregamos al programa para indicar el objeto o valor con el que debe trabajar. Según el comando, puede ser una ruta, un nombre, un número o una URL. En este ejemplo, `labs` dirige la operación hacia esa carpeta; si ejecutaras solamente `ls`, el programa listaría la carpeta actual.

Las opciones cortas suelen comenzar con un guion (`-l`) y muchas pueden combinarse (`-lh`). Las opciones largas suelen comenzar con dos guiones (`--version`) y son más fáciles de leer. Cada programa define sus propias opciones y argumentos: `-h` no significa lo mismo en todos los comandos, y no todos los comandos requieren un argumento.

Atajos útiles: **Tab** completa nombres y reduce errores; las flechas arriba/abajo recorren el historial; **Ctrl+C** interrumpe un comando que sigue ejecutándose sin cerrar la terminal.

## 3. 🧭 Rutas absolutas, relativas y profundidad

Una **ruta** o *path* es la dirección de un archivo o carpeta. Cada nombre separado por `/` representa un nivel. La carpeta desde la que el shell interpreta los comandos se llama **directorio actual** y se consulta con `pwd` (*print working directory*).

### La raíz `/`, la carpeta personal `~` y el directorio actual `.`

Estos tres símbolos no son equivalentes:

- `/` representa la raíz del sistema de archivos que ve el shell;
- `~` representa tu carpeta personal o *home*;
- `.` representa la carpeta en la que estás ahora mismo.

Muchos shells se abren en `~`, por eso el prompt suele mostrar ese símbolo. El shell lo expande a una ruta distinta para cada persona:

```text
/Users/ana                  # ~ en macOS
/home/ana                   # ~ en Linux
/c/Users/Ana                # ~ típico en Git Bash de Windows
```

En Git Bash, una ruta de Windows como `C:\Users\Ana` se representa como `/c/Users/Ana`. No necesitas memorizarla: `cd ~` te lleva al *home* y `pwd` muestra su ruta absoluta.

```bash
cd ~
pwd
```

### Ruta absoluta y ruta relativa

Una **ruta absoluta** comienza en la raíz y señala el mismo lugar sin importar el directorio actual. Una **ruta relativa** parte del directorio actual; normalmente no empieza con `/` y su resultado cambia según dónde estés.

Supongamos que abriste en VS Code el repositorio público y la terminal está en su raíz. El árbol siguiente muestra la profundidad. La estructura bajo `cursos/` es sólo un ejemplo; lo importante son las relaciones entre niveles:

```text
~/                                             # carpeta personal
└── cursos/                                    # nivel 1
    └── proyecto-ciencia-datos/                # nivel 2: carpeta común
        ├── proyecto-ciencia-datos-2026-2/     # nivel 3 ← estás aquí
        │   └── labs/                          # nivel 4
        │       └── starters/                  # nivel 5
        │           └── clase-02-python-vscode/ # nivel 6
        ├── pcd-entregas-2026/                 # aparecerá después de clonar
        └── proyecto/
```

Desde esa ubicación, una ruta absoluta al starter podría ser:

```text
/Users/ana/cursos/proyecto-ciencia-datos/proyecto-ciencia-datos-2026-2/labs/starters/clase-02-python-vscode
```

La ruta relativa equivalente es mucho más corta porque comienza en la raíz del repositorio actual:

```text
labs/starters/clase-02-python-vscode
```

### Subir y bajar varios niveles

Cada `..` sube exactamente un nivel. Los segmentos unidos por `/` permiten recorrer varios niveles con un solo comando:

| Ruta ejecutada desde el repositorio del curso | Destino |
|---|---|
| `.` | `proyecto-ciencia-datos-2026-2/` |
| `..` | carpeta común `proyecto-ciencia-datos/` |
| `../..` | carpeta `cursos/` del ejemplo |
| `../../..` | carpeta personal `~` del ejemplo |
| `labs/starters` | baja dos niveles dentro del repositorio |
| `labs/starters/clase-02-python-vscode` | baja tres niveles hasta el starter |

También puedes combinar una subida con varios descensos. Más adelante, desde `pcd-entregas-2026`, esta ruta llegará al starter que está en el repositorio hermano:

```text
../proyecto-ciencia-datos-2026-2/labs/starters/clase-02-python-vscode
```

Se lee de izquierda a derecha: sube a la carpeta común con `..`, entra al repositorio público y baja por `labs`, `starters` y la carpeta de la clase.

### Inspeccionar una ruta sin cambiar de carpeta

`cd ruta` cambia el directorio actual. `ls ruta` sólo muestra el contenido del destino y después permaneces donde estabas. Desde la raíz del repositorio público prueba:

```bash
pwd
ls ..
ls labs/starters/clase-02-python-vscode
ls ../../..
pwd
```

Los dos `pwd` deben mostrar la misma ruta: los comandos `ls` inspeccionaron un nivel superior, tres niveles inferiores y varios niveles superiores sin mover el shell.

`ls -R ruta` usa la flag `-R`, que significa **recursivo**, para listar también los niveles contenidos debajo de una carpeta:

```bash
ls -R labs/starters/clase-02-python-vscode
```

Úsala sólo sobre una carpeta pequeña y conocida; en una ruta grande puede producir cientos de líneas.

### Símbolos útiles en una ruta

| Símbolo | Significado | Ejemplo |
|---|---|---|
| `.` | directorio actual | `ls .` |
| `..` | directorio superior | `cd ..` |
| `~` | carpeta personal | `cd ~` |
| `-` | directorio anterior visitado con `cd` | `cd -` |
| `/` | separa niveles | `labs/starters` |

Los espacios separan argumentos. Si una ruta existente contiene espacios, enciérrala entre comillas, por ejemplo `cd "Proyecto en Ciencia de Datos"`. Para rutas nuevas recomendamos nombres sin espacios, acentos ni caracteres especiales. Usa **Tab** para completar cada segmento y confirma el resultado con `pwd` y `ls`.


## 4. Comandos esenciales de CLI

No se busca memorizar la tabla en una sesión. La meta es reconocer familias de acciones, leer el comando antes de ejecutarlo y consultar ayuda cuando sea necesario.

### Opciones o *flags* y cómo consultar ayuda

Una *flag* modifica la manera en que se ejecuta un comando. En `ls -la`, `-l` solicita una lista detallada y `-a` incluye nombres ocultos; Bash permite agruparlas como `-la`. En `mkdir -p`, `-p` crea los directorios intermedios que falten. En `head -n 20 archivo.py`, `-n` necesita el valor `20`: no todas las flags funcionan solas. Las opciones largas se escriben completas, como `git branch --show-current` o `python --version`.

No adivines una flag. Consulta la ayuda del propio programa:

```bash
git switch -h          # resumen breve dentro de la terminal
git help -w switch     # documentación detallada de Git en el navegador
python --help          # opciones generales de Python
```

Estos tres ejemplos son las opciones comunes que utilizaremos en Git Bash de Windows, Zsh de macOS y Bash de Linux. `-h` muestra una referencia breve; `git help -w` usa la flag `-w` (*web*) para abrir la documentación detallada en el navegador; `--help` solicita la ayuda de Python. La sección **Usage** o **Uso** muestra el orden esperado y los elementos entre corchetes suelen ser opcionales.

También existen ayudas que dependen del entorno. `man ls` es habitual en macOS y Linux, pero **no debemos asumir que `man` está instalado en Git Bash**. Por otro lado, `ls --help` funciona en Git Bash y en sistemas Linux con herramientas GNU, pero el `ls` incluido en macOS no ofrece esa misma flag. Por eso, antes de copiar una alternativa de internet, identifica tu sistema y prioriza la ayuda propia del programa que estás usando.

| Comando | Qué hace | Ejemplo |
|---|---|---|

| `ls` | lista contenido de la carpeta actual o indicada | `ls actividades` |

| `ls -a` | `-a`: incluye nombres ocultos como `.git` | `ls -a` |
| `ls -lh` | combina `-l` con `-h`: tamaños legibles para personas | `ls -lh` |
| `ls -R ruta` | `-R`: lista recursivamente los niveles contenidos en una ruta | `ls -R actividades` |

### Crear, copiar y mover

| Comando | Qué hace | Ejemplo |
|---|---|---|
| `mkdir nombre` | crea una carpeta | `mkdir actividades` |

### Leer, buscar y contar

| Comando | Qué hace | Ejemplo |

## 5. 🧠 Teoría de Git: del archivo local al repositorio remoto

### ¿Qué problema resuelve Git?

**Git** es un sistema de control de versiones: registra cambios significativos de un proyecto para consultar qué cambió, quién lo cambió, cuándo y por qué. También permite desarrollar líneas de trabajo separadas y recuperar versiones anteriores. Git funciona en tu equipo y puede crear historial sin conexión a internet.

**GitHub** no es Git. Es un servicio que aloja repositorios remotos y agrega colaboración mediante pull requests, revisiones y permisos. Podemos usar Git sin GitHub; en el curso utilizaremos ambos.

### Cómo leer un comando de Git

Git reúne varias acciones dentro del mismo programa. Por eso sus comandos suelen seguir esta estructura:

```text
git  subcomando  opciones  argumentos
```

- `git` es el programa que ejecuta el shell;
- el **subcomando** indica la acción, por ejemplo `status`, `branch`, `switch`, `add` o `commit`;
- las **opciones** modifican esa acción, como `--show-current` o `-m`;
- los **argumentos** aportan los datos necesarios, como una ruta, un nombre de rama o un mensaje.

Por ejemplo, en `git branch --show-current`, `branch` es el subcomando y `--show-current` es la opción que pide mostrar únicamente la rama activa. Cada subcomando acepta sus propias opciones y argumentos; no se pueden intercambiar libremente.

### Repositorio y directorio de trabajo

Un **repositorio** o *repo* es un proyecto administrado por Git. Incluye dos partes relacionadas:

- el **directorio de trabajo**, con los archivos y carpetas que ves y editas;
- el directorio oculto `.git`, donde Git conserva commits, ramas, referencias y configuración local.

Una carpeta con archivos no es automáticamente un repositorio. `git clone` crea una copia local de un repositorio existente, descarga su historial y configura la conexión con el repositorio de origen.

### Staging y commit

Un **commit** es un punto confirmado del historial: registra una instantánea de las versiones seleccionadas, junto con identificador, autoría, fecha, mensaje y vínculo con el commit anterior.

Guardar en VS Code y crear un commit son acciones distintas:

1. guardas un archivo en el directorio de trabajo;
2. `git add` selecciona su versión actual en el **área de staging**;
3. `git commit` registra esa selección en el repositorio local.

Ninguna de estas acciones publica todavía en GitHub. El staging permite revisar y agrupar una decisión concreta sin incluir automáticamente todos los archivos modificados.

### Ramas y rama activa

Una **rama** (*branch*) es un nombre móvil que señala un commit del historial. `main` es la rama principal del repositorio. Al crear otra rama desde `main`, ambas comienzan señalando el mismo commit; cuando registras cambios nuevos, sólo avanza la rama activa. `main` permanece en su posición hasta que los cambios se integran.

```text
A---B  main
     \
      C---D  feat/resumen-viajes
```

En este esquema, `feat/resumen-viajes` nació en `B` y avanzó con los commits `C` y `D`; `main` todavía señala `B`. Una rama no es una segunda copia física de toda la carpeta: es una referencia ligera dentro del historial.

Git denomina **rama activa** a la rama sobre la que estás trabajando. Los commits nuevos avanzan esa rama y los archivos del directorio de trabajo corresponden a la versión que seleccionaste. Estos son los comandos básicos que utilizaremos:

| Comando | Uso |
|---|---|
| `git branch` | Lista las ramas locales; `*` marca la activa. |
| `git branch --show-current` | Muestra únicamente el nombre de la rama activa. |
| `git switch -c nombre` | Crea una rama desde el commit actual y cambia a ella. |
| `git switch nombre` | Cambia a una rama que ya existe. |

El flujo básico será: actualizar `main`, crear una rama para un cambio concreto, trabajar y crear commits allí, publicar la rama con `git push` y proponer su integración mediante un pull request. Una rama creada con `git switch -c` existe sólo en tu repositorio local hasta que la publicas.

### Repositorio local, remoto y `origin`

El **repositorio local** vive en tu computadora. Un **repositorio remoto** es otra copia accesible por red; en este curso estará alojada en GitHub. `origin` es el nombre corto que Git asigna normalmente al remoto desde el cual clonaste. No significa “GitHub” de manera universal: es un alias configurable.

- `git pull` trae e integra en la rama local actual los cambios de su rama remota;
- `git push` publica en el remoto los commits de tu rama local;
- un **pull request** es una solicitud en GitHub para revisar y fusionar una rama con otra; no es un commit ni el comando `git pull`.

Observa el flujo completo. Primero sigue las transiciones dentro de tu equipo y después identifica cuáles cruzan hacia GitHub:

![Diagrama del flujo entre directorio de trabajo, staging, repositorio local y repositorio remoto en GitHub](../assets/modulo-01-fundamentos/clase-02/git-flujo-local-remoto.svg)

La idea esencial es: **guardar no prepara, preparar no crea un commit y crear un commit no lo publica**.


## 6. 🛠️ Preparar y clonar el repositorio de entregas

Ahora aplicaremos el modelo anterior. Ejecuta cada bloque por separado y lee la salida antes de continuar. Si una ruta o un repositorio no coincide con lo esperado, detente y pide ayuda.

### A. Confirmar y actualizar el repositorio público

Abre en VS Code `proyecto-ciencia-datos-2026-2`. En Windows selecciona Git Bash como perfil de la terminal integrada:

```bash
pwd
ls -la
git branch --show-current
git status
```

- `pwd` debe terminar en `proyecto-ciencia-datos-2026-2`;
- `ls -la` debe mostrar `.git`; la flag `-a` incluye nombres ocultos;
- `git branch --show-current` debe mostrar `main`;
- `git status` indica la rama y cualquier cambio local.

Si `git status` no muestra modificaciones que desconozcas, actualiza el material:

```bash
git pull
```

`git pull` descarga los commits nuevos de `origin` y los integra en la rama local actual. Si aparecen archivos modificados que no reconoces, **no ejecutes pull ni comandos de limpieza**: conserva la salida de `git status` y pide ayuda.

### B. Crear el repositorio privado en GitHub

En la interfaz web de GitHub:

1. selecciona **New repository**;
2. escribe `pcd-entregas-2026`;
3. selecciona **Private**;
4. activa **Add a README file** para que la rama `main` exista desde el inicio;
5. crea el repositorio;
6. entra a **Settings > Collaborators** y agrega a `zapatacc`.

El repositorio remoto ya existe en GitHub, pero todavía no hay una copia en tu computadora. En la página del repositorio, selecciona **Code > HTTPS** y copia su URL. Tendrá una forma parecida a `https://github.com/TU-USUARIO/pcd-entregas-2026.git`.

### C. Ir a la carpeta común y clonar exactamente allí

Regresa a la terminal que está en la raíz del repositorio público. Como ambos repositorios deben quedar al mismo nivel, `cd ..` sube a su carpeta común:

```bash

Usaremos `feat` para una capacidad nueva, `fix` para corregir un error, `docs` para documentación, `test` para verificaciones, `refactor` para reorganizar sin cambiar comportamiento y `chore` para mantenimiento. ```text
feat(viajes): agrega estimacion de duracion
test(viajes): verifica distancias invalidas
docs(tarea-01): compara definiciones de mlops
```

Conventional Commits define **mensajes de commit**, no nombres de ramas. La semejanza entre ambos es una convención adicional del curso para que la intención sea reconocible. Evita mensajes como `cambios`, `tarea`, `final`, `ahora si` o `version 2`.

## 8. 🐍 Python incremental: estimar duración de viajes

Retomaremos la función de la clase 1. La primera versión sólo recibe la distancia, calcula 4 minutos por kilómetro más 2 minutos fijos y redondea a **una cifra decimal**. Los datos son ficticios y la regla es didáctica, no un modelo real.

Antes de ejecutar, observa tres elementos de sintaxis que utilizaremos:

- En `distancia_km: float`, `: float` documenta que esperamos un número decimal. En `-> float`, la flecha documenta el tipo de resultado esperado. Estas **anotaciones de tipo** ayudan a leer y revisar el código, pero Python no convierte ni valida automáticamente los valores sólo por escribirlas.
- Un solo signo `=` **asigna** un valor, como en `duracion = 14.8`. Dos signos `==` **comparan** dos valores y producen `True` o `False`, como en `resultado == 14.8`.
- `assert condicion` es una verificación rápida: si la condición es `True`, el programa continúa; si es `False`, se detiene con `AssertionError`. Aquí lo usaremos para comprobar ejemplos conocidos, no para validar datos ingresados por una persona usuaria.

Por eso `assert estimar_duracion_base(3.2) == 14.8` significa: “ejecuta la función y confirma que su resultado sea exactamente `14.8`”.


````python
def estimar_duracion_base(distancia_km: float) -> float:
    """Estima una duración base y redondea a una cifra decimal."""
    duracion = distancia_km * 4 + 2
    return round(duracion, 1)


assert estimar_duracion_base(3.2) == 14.8
print(estimar_duracion_base(3.2))
````

La nueva versión recibe también pasajeros y fin de semana. Debe validar que la distancia sea positiva, sumar 3 minutos si viajan más de dos personas y reducir 10 % en fin de semana. Después `resumir_viajes` recorre una lista de diccionarios importada desde `datos_viajes.py`.

En el starter aparecerán algunas construcciones adicionales:

- `from datos_viajes import VIAJES` carga desde `datos_viajes.py` la lista llamada `VIAJES`. Las mayúsculas comunican por convención que usaremos esos datos como una constante de ejemplo.
- `raise ValueError("mensaje")` detiene la función e informa que el valor recibido no es válido; por ejemplo, una distancia negativa.
- Una cadena que comienza con `f`, como `f"{origen} → {destino}"`, es una **f-string**: evalúa lo escrito entre llaves e inserta el resultado en el texto.
- `list[dict]` es otra anotación: comunica que esperamos una lista cuyos elementos sean diccionarios.

````python
viaje = {
    "origen": "Centro",
    "destino": "Chapultepec",
    "distancia_km": 3.2,
    "pasajeros": 1,
    "fin_de_semana": False,
}

print(viaje["origen"], "→", viaje["destino"])
print(estimar_duracion_base(viaje["distancia_km"]))
````

## 10. 🧾 Registrar los cambios en Git local

El programa ya funciona; ahora convertirás esos cambios guardados en uno o más commits comprensibles. Sigue el mismo ciclo cada vez: **inspeccionar → preparar → revisar staging → crear commit → comprobar**.

### A. Inspeccionar el directorio de trabajo

```bash
git status
git branch --show-current
git diff -- actividades/clase-02
```

- `git status` resume la rama, archivos nuevos, modificaciones y staging;
- `git branch --show-current` confirma que sigues en `feat/resumen-viajes`;
- `git diff` muestra cambios no preparados en archivos que Git ya conoce.

Los dos guiones solos (`--`) separan opciones de rutas y aclaran que `actividades/clase-02` es una ruta. Como los archivos copiados son nuevos y todavía están *untracked*, `git diff` puede no mostrar su contenido; `git status` sí los reportará.

### B. Preparar y revisar staging

```bash
git add actividades/clase-02
git status
git diff --staged
```

`git add` no publica ni crea un commit: selecciona la versión actual de esos archivos para el próximo commit. `git diff --staged` compara lo preparado contra el último commit y permite revisar exactamente qué se registrará.

Si editas otra vez después de `git add`, ejecuta `git status`: puede existir una versión preparada y otra todavía sin preparar. Vuelve a usar `git add` sólo cuando quieras actualizar la selección.

Si Git solicita identidad, configúrala únicamente en este repositorio:

```bash
git config user.name "Tu Nombre"
git config user.email "tu-correo-asociado-a-github"
```

Estos valores identifican la autoría del commit; no son las credenciales para iniciar sesión en GitHub.

### C. Crear y revisar el commit

Crea el commit sólo después de comprender `git diff --staged`. La forma que utilizaremos es:

```bash
git commit -m "feat(viajes): agrega estimacion de duracion"
```

| Parte | Papel |
|---|---|
| `git` | Ejecuta el programa Git. |
| `commit` | Crea un commit local con la versión que está en staging. No incluye automáticamente archivos sin preparar. |
| `-m` | Es la forma corta de `--message`; indica que el siguiente argumento será el mensaje del commit. |
| `"feat(viajes): agrega estimacion de duracion"` | Es el argumento de `-m`. Las comillas hacen que el shell trate el texto con espacios como un solo mensaje. |

El mensaje queda guardado en el historial y debe explicar la decisión registrada; no es el nombre de un archivo ni un comentario que se envía a GitHub. Si omites `-m`, Git abre un editor para solicitar el mensaje. En esta primera práctica usaremos `-m` para mantener visible el comando completo. Crear el commit tampoco publica nada: eso ocurrirá después con `git push`.

Cuando el commit se crea correctamente, Git muestra la rama activa, un identificador corto y un resumen de archivos modificados. Después revisa el resultado:

```bash
git status
git show --stat HEAD
git log --oneline --graph --decorate --all -5
```

- `git status` debe indicar que no quedan cambios pendientes;
- `HEAD` representa el commit actualmente seleccionado y `--stat` muestra un resumen de archivos y líneas;
- en `git log`, `--oneline` usa una línea por commit, `--graph` dibuja el historial, `--decorate` agrega nombres de ramas, `--all` incluye las ramas locales conocidas y `-5` limita la salida a cinco commits.

Un commit debe representar una decisión comprensible. Si después agregas verificaciones o corriges un error real, repite `git add`, revisa staging y crea otro commit, por ejemplo:

```bash
git commit -m "test(viajes): verifica casos principales"
```

No fabriques commits vacíos sólo para aumentar la cantidad.

**✅ Checkpoint:** `git status` no muestra cambios pendientes y `git log -1 --oneline` presenta un mensaje que explica el cambio.


## 11. 🔀 Publicar, documentar y fusionar el pull request

Un **pull request** es una propuesta para revisar e integrar los commits de una rama en otra. Se crea en GitHub; no es lo mismo que `git pull`. `origin` es el nombre convencional del repositorio remoto creado por `git clone`. `push` envía commits locales; no sustituye al commit. Comprueba el destino y publica la rama:

```bash
git remote -v
git push -u origin feat/resumen-viajes
```

La opción `-u` es la forma corta de `--set-upstream`: conecta la rama local con `origin/feat/resumen-viajes`. Después de la primera publicación, Git ya conoce el destino y los nuevos commits de esa rama pueden enviarse con `git push`.

### La plantilla de pull request

La plantilla canónica está en `docs/plantillas/pull-request-tarea.md` dentro del **repositorio público del curso**. No está en `.github/` dentro de tu repositorio privado y GitHub no la insertará automáticamente. El archivo publicado se conserva en `docs/plantillas/` para que siempre exista una copia limpia disponible para ésta y las tareas futuras.

Para usarla, abre `proyecto-ciencia-datos-2026-2/docs/plantillas/pull-request-tarea.md` en VS Code o en GitHub, selecciona y copia **todo** su contenido. Después pégalo en **Write**, dentro de la descripción del nuevo PR. Sustituye las instrucciones por información concreta, pero conserva sus secciones y checklist. Es texto Markdown: `##` crea encabezados, los acentos graves marcan código y `- [ ]` crea casillas; cambia a `- [x]` sólo lo que ya cumpliste. Usa **Preview** para ver cómo se renderizará antes de crear o fusionar el PR.

### Crear el PR desde la interfaz de GitHub

1. abre el repositorio privado en GitHub y selecciona **Compare & pull request** o **Pull requests > New pull request**;
2. confirma **base: main** y **compare: feat/resumen-viajes**; la flecha debe llevar los cambios de tu rama hacia `main`;
3. escribe un título descriptivo, por ejemplo `feat(viajes): agrega estimacion de duracion`;
4. abre la plantilla del repositorio público, copia todo su contenido y pégalo en **Write**; completa tarea, ruta, resumen y verificación;
5. marca sólo las casillas que realmente cumpliste y usa **Preview** para revisar el Markdown;
6. selecciona **Create pull request**; crear el PR todavía no hace merge;
7. abre **Files changed** y confirma que sólo aparezcan los archivos de `actividades/clase-02/`; la plantilla está en la descripción y no debe aparecer como archivo modificado;
8. si detectas un error, corrígelo localmente, crea otro commit y ejecuta `git push`; el PR se actualizará automáticamente;
9. selecciona **Merge pull request > Create a merge commit** y confirma;
10. verifica que el PR muestre la etiqueta morada **Merged** y esté cerrado.

El merge modifica el `main` remoto. Tu `main` local todavía debe actualizarse:

```bash
git switch main
git pull
git log --oneline --graph --decorate -5
```

`git switch main` cambia la rama local; `git pull` trae el merge realizado en GitHub. Cerrar un PR sin merge no incorpora el trabajo a `main`. Dejarlo abierto tampoco completa el flujo. En este curso toda tarea debe terminar con un PR cerrado **y fusionado** a `main`.
