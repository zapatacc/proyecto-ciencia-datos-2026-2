## 1. Crear el repositorio del equipo

Una persona crea un repositorio nuevo en GitHub:

1. elijan un nombre corto, descriptivo, en minúsculas y con guiones, por ejemplo `proyecto-demanda-bicicletas`;
2. seleccionen visibilidad **Private**;
3. inicialícenlo con un `README.md`;
4. agreguen como colaboradores a los demás integrantes;
5. agreguen como colaborador al usuario de GitHub **`zapatacc`**;
6. esperen a que todos acepten la invitación.

Los equipos serán de dos personas, con un máximo de tres integrantes. No incluyan matrículas, correos, tokens ni credenciales en el repositorio.

**Checkpoint:** todas las personas del equipo y `zapatacc` tienen acceso al repositorio privado.

## 2. Clonar el repositorio y abrir la rama de clase

Cada integrante clona el repositorio en su computadora. Desde la carpeta que contiene los demás repositorios del curso:

```bash
git clone URL_DEL_REPOSITORIO
cd NOMBRE_DEL_REPOSITORIO
pwd
git status
git remote -v
```

Sustituye los valores en mayúsculas por la URL y el nombre reales. `pwd` confirma la ruta y `git remote -v` permite comprobar que `origin` apunta al repositorio del equipo.

Una persona crea la rama del trabajo en clase:

```bash
git switch main
git pull
git switch -c docs/trabajo-clase-07
```

En Windows utilicen Git Bash. En macOS o Linux pueden usar Zsh o Bash.

## 3. Crear la estructura inicial

Desde la raíz del repositorio ejecuten:

```bash
uv python pin 3.12
uv init --bare --python 3.12
uv add ipykernel
mkdir -p informe notebooks data src
touch data/README.md
```

`ipykernel` permite que VS Code ejecute notebooks con el ambiente del proyecto. `uv add` lo declara en `pyproject.toml`, actualiza `uv.lock` y crea o sincroniza `.venv/`.

Después creen `informe/informe-final.ipynb` desde VS Code:

1. abran la carpeta `informe/`;
2. creen un archivo llamado `informe-final.ipynb`;
3. seleccionen el kernel de `.venv/` cuando VS Code lo solicite.

La estructura inicial será:

```text
.
├── .gitignore
├── .python-version
├── README.md
├── pyproject.toml
├── uv.lock
├── data/
│   └── README.md
├── informe/
│   └── informe-final.ipynb
├── notebooks/
└── src/
```

`notebooks/` guardará análisis y experimentos detallados. `src/` guardará posteriormente el código reutilizable. Git mostrará esas carpetas cuando contengan su primer archivo.

## 4. Preparar `.gitignore`

Creen `.gitignore` en la raíz:

```gitignore
.venv/
__pycache__/
*.py[cod]
.DS_Store
.env
artifacts/
data/raw/
data/interim/
data/processed/
```

Estas reglas mantienen fuera de Git ambientes, secretos, datos de trabajo y modelos generados. Antes de cada commit revisen siempre `git status`; `.gitignore` no sustituye esa comprobación.

## 5. Cómo crecerá el repositorio

Durante el semestre se agregarán archivos conforme se necesiten:

```text
.
├── README.md
├── informe/
│   └── informe-final.ipynb
├── notebooks/
│   ├── 01-eda.ipynb
│   ├── 02-preparacion-datos.ipynb
│   └── 03-experimentos.ipynb
├── data/
│   └── README.md
└── src/
    ├── entrenamiento/
    ├── api/
    └── interfaz/
```

El **informe final** sintetizará el proyecto. No debe duplicar todo el código de los notebooks o de `src/`; debe explicar los hallazgos y enlazar el archivo donde vive el detalle técnico.

## 7. Elegir el dataset

Elijan un dataset relacionado con el problema del equipo. Debe cumplir como mínimo:

- fuente pública o acceso permitido;
- licencia o condiciones de uso identificables;
- una muestra significativa para el problema que quieren estudiar;
- variables suficientes para analizar y construir un modelo;
- una variable objetivo identificable o construible;
- tamaño manejable en las computadoras del equipo;
- ausencia de datos personales o sensibles que no sean necesarios;
- relación clara con el problema planteado.

Abran una muestra y confirmen qué representa una fila, cuáles podrían ser las features y cuál sería el target. No elijan únicamente por el número de filas o porque el dataset aparece en una lista de recomendaciones.

### Fuentes para buscar datasets

Pueden comenzar en estos catálogos:

- [UCI Machine Learning Repository](https://archive.ics.uci.edu/): conjuntos preparados para problemas de machine learning.
- [Kaggle Datasets](https://www.kaggle.com/datasets): catálogo amplio; revisen la fuente original y la licencia de cada publicación.
- [OpenML](https://www.openml.org/search?type=data): permite filtrar datasets y consultar sus características.
- [Google Dataset Search](https://datasetsearch.research.google.com/): buscador de datasets publicados en distintos sitios; deben visitar y verificar al proveedor original.
- [INEGI — Descarga masiva](https://www.inegi.org.mx/app/descarga/default.html): información estadística y geográfica de México.
- [Data.gov](https://data.gov/): catálogo de datos abiertos del gobierno de Estados Unidos.
- [World Bank Open Data](https://data.worldbank.org/): indicadores y datos públicos de desarrollo internacional.
- [NYC Open Data](https://opendata.cityofnewyork.us/): datos públicos de la ciudad de Nueva York.

El catálogo ayuda a descubrir datos, pero el equipo debe registrar en `data/README.md` la página del dataset específico, su responsable, su licencia y el corte utilizado.

### Documentar los datos

Completen `data/README.md`:

```markdown
# Datos del proyecto

## Fuente

URL de la fuente original y organización responsable.

## Licencia o condiciones de uso

Indica el permiso identificado y enlaza su evidencia.

## Descripción

Explica qué representa una fila, el tamaño aproximado y las columnas principales.

## Target y features iniciales

Indica qué se quiere predecir y qué información podría utilizarse.

## Obtención de los datos

Explica cómo descargar la versión o corte utilizado.

## Archivos locales

Indica qué archivos permanecen fuera de Git.
```

No agreguen el dataset completo al repositorio. Documenten cómo obtenerlo y manténganlo bajo `data/raw/`, que está ignorado por Git.

## Errores frecuentes

- **Un integrante no puede acceder:** revisar la invitación y la cuenta de GitHub utilizada.
- **El repositorio se creó sin README:** crear el primer commit en `main` antes de abrir la rama de entrega.
- **`uv` utiliza otra versión:** ejecutar `uv python find 3.12`; si no existe, usar `uv python install 3.12`.
- **El dataset no tiene licencia clara:** no asumir permiso; buscar la fuente original o elegir otro dataset.
- **El target contiene información del futuro:** revisar el riesgo de fuga antes de continuar.
- **Los datos aparecen en `git status`:** comprobar que estén bajo `data/raw/` y revisar `.gitignore`.
- **El equipo no puede explicar el problema en una frase:** regresar a contexto, usuario, decisión y resultado esperado.
