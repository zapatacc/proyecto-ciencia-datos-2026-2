## Descripción de la asignatura

En ciencia de datos, el desafío no termina al desarrollar y evaluar un modelo predictivo. Para generar valor, la solución debe poder ejecutarse, compartirse e integrarse con otras aplicaciones o personas. También debe ser posible identificar qué versión está en uso, observar su comportamiento y recuperarse de un cambio defectuoso.

Este curso desarrolla las competencias necesarias para llevar soluciones de ciencia de datos a un entorno de producción profesional mediante prácticas de desarrollo de software y tecnologías de información. El énfasis está en construir un sistema pequeño y coherente, no en realizar una competencia extensa de modelado.

El caso conductor será **NYC Taxi**. Una estimación de duración de viaje evolucionará desde una función Python hasta una aplicación desplegada y operada de forma básica.

## Objetivo general

> Desarrollar la capacidad de convertir una solución de ciencia de datos en un sistema que otras personas puedan ejecutar, consumir, verificar y mantener.

Al finalizar el semestre, el estudiantado relacionará código, datos y modelo; expondrá predicciones mediante una API; rastreará y versionará experimentos; conectará una interfaz; construirá contenedores; publicará una aplicación y demostrará acciones básicas de operación como consultar logs, actualizar o regresar a una versión anterior.

La versión estudiantil de la guía está en `docs/guia-de-aprendizaje.md`.

## Temas y subtemas

El recorrido se organiza en tres módulos progresivos.

### Módulo 1 — Fundamentos para construir y compartir aplicaciones de datos

1. Bienvenida, guía de aprendizaje y herramientas.
2. Python en VS Code.
3. Git Bash, terminal y Git local.
4. GitHub, ambientes y dependencias.
5. Cliente, servidor, HTTP y JSON.
6. Primera aplicación FastAPI.
7. Primera API de predicción.

**Resultado del módulo:** exponer localmente una predicción mediante FastAPI.

### Módulo 2 — Del modelo al producto: diseño y trazabilidad MLOps

1. Fundamentos y niveles de madurez de MLOps.
2. Datos, modelo y código.
3. Patrones de despliegue y diseño del proyecto.
4. Experiment tracking con MLflow local y Databricks.
5. Model Registry, versiones y aliases.
6. Pipelines de ML.
7. Model as a Service y Streamlit.

**Resultado del módulo:** rastrear, registrar, cargar y consumir el modelo canónico.

### Módulo 3 — De local a producción: contenedores, nube y operación

1. Introducción a Docker y contenerización de la API.
2. Redes y Docker Compose.
3. Streamlit en Hugging Face Docker Spaces.
4. Fundamentos de cloud computing.
5. Introducción guiada a AWS.
6. Logs, actualización, rollback y cierre de recursos.
7. Demostración y defensa final.

**Resultado del módulo:** ejecutar la aplicación con contenedores, publicarla y demostrar acciones básicas de operación.

## Cierre — preparación del espacio de trabajo

Prepararemos una carpeta local para trabajar durante todo el semestre. Una organización posible es tener, como elementos separados, el repositorio de materiales del curso, las tareas y el proyecto. Es sólo una sugerencia: puedes elegir la ubicación y los nombres que te resulten más cómodos.

> **Importante:** no clones este repositorio dentro de OneDrive. Las carpetas sincronizadas pueden bloquear archivos, cambiar rutas o producir conflictos con Git. Usa una ruta local; por ejemplo, `C:\Users\tu_usuario\cursos` en Windows, `/Users/tu_usuario/cursos` en macOS o `/home/tu_usuario/cursos` en Linux.

### 1. Instalar las herramientas que falten

- [Git](https://git-scm.com/downloads/) — en Windows, Git for Windows también instala Git Bash.
- [Visual Studio Code](https://code.visualstudio.com/download) y su extensión oficial de Python.
- [Python 3.12](https://www.python.org/downloads/release/python-31210/) — en Windows, habilita la opción para agregar Python al PATH.

Abre Git Bash en Windows o una terminal en macOS/Linux y verifica la instalación:

```bash
git --version
bash --version
python --version     # usa python3 si tu sistema lo requiere
```

### 2. Crear las carpetas desde la interfaz

Usa el **Explorador de archivos** en Windows o **Finder** en macOS y elige una ubicación local fácil de encontrar. Como ejemplo, puedes crear `cursos`, dentro `proyecto-ciencia-datos` y, dentro de esta última, las carpetas `tareas` y `proyecto`. Revisa la barra de dirección del Explorador o Finder: la ruta no debe contener `OneDrive`. Las carpetas `tareas` y `proyecto` son para tu trabajo; no guardes esos archivos dentro del repositorio de materiales salvo que una actividad lo indique expresamente.

- Python está instalado, pero no aparece en PATH.
- En Windows se abrió PowerShell en lugar de Git Bash.
- La ruta contiene `OneDrive` u otra carpeta sincronizada.
- `git clone` se ejecutó fuera de la carpeta local elegida para el curso.
- VS Code sigue mostrando la carpeta principal: usa **Archivo > Abrir carpeta...** y selecciona la carpeta clonada.
- El archivo se modificó, pero no se guardó antes de ejecutarlo.
- Una instalación está bloqueada por una política institucional.
